## CIFAR 10 Image Classification

```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
```

```python
cifar_10=keras.datasets.cifar10
(x_train, y_train), (x_test, y_test)=cifar_10.load_data()
```

```python
print(x_train.shape)
```

```python
print(x_test.shape)
```

```python
class_names=['airplane', 'automobile', 'bird', 'cat', 'deer',
             'dog', 'frog', 'horse', 'ship', 'truck']
```

```python
y_train_flat=y_train.reshape(-1,)
```

```python
print(y_train_flat)
```

```python
def plot_sample(x, y, idx):
    plt.figure(figsize=(4, 4))
    plt.imshow(x[idx], interpolation='nearest')
    plt.xlabel(class_names[y[idx]])
```

```python
plot_sample(x_train, y_train_flat, 3)
```

```python
# Normalize the values
x_train=x_train/255
x_test=x_test/255
```

```python
model=keras.models.Sequential([
    keras.Input(shape=(32, 32, 3)),
    keras.layers.Flatten(),
    keras.layers.Dense(units=3000, activation='relu'),
    keras.layers.Dense(units=1000, activation='relu'),
    keras.layers.Dense(units=10, activation='sigmoid'),
])
```

```python
model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='sgd',
    metrics=['accuracy']
)
```

```python
model.fit(x_train, y_train, epochs=20)
```

```python
model.evaluate(x_test, y_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

y_pred=model.predict(x_test)
y_pred_classes=[np.argmax(ele) for ele in y_pred]
print(f'Classification Report:\n{classification_report(y_test, y_pred_classes)}')
```

### CNN

```python
cnn_model=keras.models.Sequential([
    # cnn
    keras.Input(shape=(32, 32, 3)),
    keras.layers.Conv2D(filters=32, kernel_size=(3, 3), activation='relu'),
    keras.layers.MaxPooling2D((2, 2)),

    keras.layers.Conv2D(filters=64, kernel_size=(3, 3), activation='relu'),
    keras.layers.MaxPooling2D((2, 2)),

    # dense
    keras.layers.Flatten(),
    keras.layers.Dense(units=64, activation='relu'),
    keras.layers.Dense(units=10, activation='softmax')
])
```

```python
cnn_model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
cnn_model.fit(x_train, y_train, epochs=15)
```

```python
cnn_model.evaluate(x_test, y_test)
```

```python
plot_sample(x_test, y_test.reshape(-1,), 1)
```

```python
y_pred_cnn=cnn_model.predict(x_test)
```

```python
y_pred_cnn[:2]
```

```python
y_classes=[np.argmax(ele) for ele in y_pred_cnn]
```

```python
y_classes[:2]
```

```python
print(class_names[y_classes[1]])
```

```python
plot_sample(x_test, y_test.reshape(-1,), 0)
```

```python
print(class_names[y_classes[0]])
```

```python
print(f'Classification Report:\n{classification_report(y_test, y_classes)}')
```

### CNN for Project

```python
from tensorflow import keras

cnn_model_proj = keras.models.Sequential([
    # First Conv Block
    keras.layers.Input(shape=(32, 32, 3)),
    keras.layers.Conv2D(64, (3,3), padding='same', activation='relu'),
    keras.layers.BatchNormalization(),
    keras.layers.Conv2D(64, (3,3), padding='same', activation='relu'),
    keras.layers.MaxPooling2D((2,2)),
    keras.layers.Dropout(0.25),

    # Second Conv Block
    keras.layers.Conv2D(128, (3,3), padding='same', activation='relu'),
    keras.layers.BatchNormalization(),
    keras.layers.Conv2D(128, (3,3), padding='same', activation='relu'),
    keras.layers.MaxPooling2D((2,2)),
    keras.layers.Dropout(0.25),

    # Dense Block
    keras.layers.Flatten(),
    keras.layers.Dense(256, activation='relu'),
    keras.layers.BatchNormalization(),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(10, activation='softmax')
])
```

```python
cnn_model_proj.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
x_valid, x_train=x_train[:5000], x_train[5000:]
y_valid, y_train=y_train[:5000], y_train[5000:]
```

```python
history=cnn_model_proj.fit(x_train, y_train, epochs=10, validation_data=(x_valid, y_valid))
```

```python
import pandas as pd

pd.DataFrame(history.history).plot(figsize=(8, 5))
plt.grid(True)
# Get the current axes
plt.gca().set_ylim(0, 1)
plt.show()
```

```python
cnn_model_proj.evaluate(x_test, y_test)
```

```python
y_pred_cnn_proj=cnn_model_proj.predict(x_test)
```

```python
y_classes_proj=[np.argmax(ele) for ele in y_pred_cnn_proj]
```

```python
print(f'Classification Report:\n{classification_report(y_test, y_classes_proj)}')
```
