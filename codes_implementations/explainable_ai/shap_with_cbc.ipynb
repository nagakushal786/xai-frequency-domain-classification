## SHAP with CatBoostClassifier

- `CatBoostClassifier` can be applied to categorical features without transforming them into one-hot encodings.

```python
import pandas as pd
import matplotlib.pyplot as plt
from catboost import CatBoostClassifier
import xgboost as xgb
from sklearn.metrics import accuracy_score, confusion_matrix
import shap

shap.initjs()
plt.rcParams.update({'figure.facecolor': 'white'})
```

```python
# Load data
data=pd.read_csv('data/mushrooms.csv')
```

```python
data.shape
```

```python
data.head()
```

```python
x=data.drop(columns='class', axis=1)
y=data['class']
y=y.astype('category').cat.codes
```

### XGBoost

```python
# Create dummy variables for the categorical features
x_dummy=pd.get_dummies(x)
```

```python
x_dummy.shape
```

```python
model=xgb.XGBClassifier()
model.fit(x_dummy, y)
```

```python
explainer=shap.Explainer(model=model)
shap_values=explainer(x_dummy)
```

```python
shap.plots.waterfall(shap_values[0])
```

### CatBoost

```python
cb_model=CatBoostClassifier(
    iterations=20,
    learning_rate=0.01,
    depth=3
)
```

```python
cat_features=list(range(len(x.columns)))
```

```python
cb_model.fit(x, y, cat_features=cat_features)
```

```python
cb_explainer=shap.Explainer(model=cb_model)
shap_values_cb=cb_explainer(x)
```

```python
shap.plots.waterfall(shap_values_cb[0])
```

```python
shap.plots.bar(shap_values_cb)
```

```python
shap.plots.beeswarm(shap_values_cb)
```
